```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
#to make it easier for analysis
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ', '_')
df = df.rename(columns={'purchase_amount_(usd)':'purchase_amount_usd'})
```

```python
df.info()
```

```python
#Create a column for age group

labels= ['Young Adult', 'Adult', 'Middle-aged', 'Senior']
df['age_group'] = pd.qcut(df['age'], q=4, labels= labels) #qcut is used to convert numerical value into categorical form
```

```python
df[['age', 'age_group']].head(10)
```

```python
#Create column puchase_frequency_days

frequency_mapping = {
    'Fortnightly' : 14,
    'Weekly': 7,
    'Monthly' : 30,
    'Quarterly' : 90,
    'Bi-Weekly' : 14,
    'Annually' : 365,
    'Every 3 Months' : 90
}

df['puchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['puchase_frequency_days', 'frequency_of_purchases']].head(10)
```

```python
# Step 1: Connect to SQL Server (SSMS) using Windows Authentication
driver = "ODBC Driver 17 for SQL Server"  
server = "localhost"                     
database = "customer_behavior"           

from sqlalchemy import create_engine
import urllib

# Format connection strings properly for SQLAlchemy
params = urllib.parse.quote_plus(
    f"DRIVER={{{driver}}};"
    f"SERVER={server};"
    f"DATABASE={database};"
    f"Trusted_Connection=yes;"
)

engine = create_engine(f"mssql+pyodbc:///?odbc_connect={params}")

# Step 2: Load DataFrame into SQL Server
table_name = "customer"      
df.to_sql(table_name, engine, if_exists="replace", index=False)

print(f"Data successfully loaded into table '{table_name}' in database '{database}'.")
```
